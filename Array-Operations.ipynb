```python
import numpy as np
```

Arithmetic operations :

```python
a1 = np.array([1,2,3,4,5])
a2 = np.array([6,7,8,9,10])
```

```python
a1 + a2
```

```python
a1 - a2
```

```python
a1 * a2
```

```python
a1 / a2
```

```python
a1 // a2
```

```python
a1 ** a2
```

Broadcasting :

```python
l = [1,2,3,4,5]
arr = np.array(l)
```

```python
arr + 10
```

Shallow And Deep Copy :

```python
a = np.arange(1,21)
a
```

```python
slice = a[:5]
slice = slice * 10
slice
```

```python
a
```

```python
b = a
b[0] = 99
b
```

Matrix Operation :

```python
A = np.array([[1,2], [3,4]])
B = np.array([[5,6], [7,8]])
```

```python
A @ B
```

```python
np.dot(A,B)
```

```python
A.T
```

```python
B.T
```

Stacking Array :

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
```

```python
np.vstack((a,b))
```

```python
np.hstack((a,b))
```

```python
np.column_stack((a,b))
```

Splitting Array :

```python
np.hsplit(c,2)
```

```python
np.vsplit(c,2)
```
